```python

"""
1. Develop a Python-based ETL script using SQLAlchemy to automate the
creation of the analytics_db schema and enforce composite primary keys on
the ratings table.
"""

from sqlalchemy import create_engine, MetaData, Table, Column, Integer, Float, String, text

DB_USER = 'username'
DB_PASS = 'password'
DB_HOST = 'localhost'

BASE_DB_URI = f"mysql+pymysql://{DB_USER}:{DB_PASS}@{DB_HOST}/"

ANALYTICS_DB_URI = f"{BASE_DB_URI}analytics_db"

def setup_analytics_database():
    base_engine = create_engine(BASE_DB_URI)
    
    with base_engine.connect() as conn:
        conn.execute(text("CREATE DATABASE IF NOT EXISTS analytics_db;"))
        conn.commit() 
        
    print("Schema 'analytics_db' has been created or already exists.")

    analytics_engine = create_engine(ANALYTICS_DB_URI)
    metadata = MetaData()

    ratings_table = Table(
        'ratings', 
        metadata,
        Column('user_id', Integer, primary_key=True, autoincrement=False),
        Column('restaurant_id', Integer, primary_key=True, autoincrement=False),
        Column('rating_score', Float, nullable=False),
        Column('review_text', String(1000))
    )
    metadata.create_all(analytics_engine)
    print("Table 'ratings' with composite primary key (user_id, restaurant_id) successfully created.")

if __name__ == "__main__":
    setup_analytics_database()
```

```python
"""
3. Write a Python validation script using Pandas to clean the Zomato dataset,
handling "NEW" or "-" values in the rating column before relational ingestion.
"""

import pandas as pd
import numpy as np

def clean_zomato_ratings(path):
    print(f"Loading dataset from {path}...")
    df = pd.read_csv(path)
    
    if 'rate' not in df.columns:
        raise ValueError("The column 'rate' was not found in the dataset.")
        
    print("Original 'rate' sample:")
    print(df['rate'].head())
    
    df['rate'] = df['rate'].replace(['NEW', '-', 'NEW ', '- '], np.nan)
    
    df['rate'] = df['rate'].astype(str).str.replace('/5', '').str.strip()
    
    df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
    
    print("\nData cleaning complete. Cleaned 'rate' sample:")
    print(df['rate'].head())
    
    
    return df

clean_zomato_ratings('C:\Users\DHRUV MODI\Documents\TOPS\Assessment\zomato.csv\zomato.csv')
```
